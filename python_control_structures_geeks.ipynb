```python
# Iterating over a list
a = [1, 2, 3]
for i in a:
    print(i)
```

```python
# Using while loop
cnt = 0
while cnt < 5:
    print(cnt)
    cnt += 1
```

```python
# Using break to exit the loop
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# Using continue to skip an iteration
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# Using pass as a placeholder
for i in range(5):
    if i == 3:
        pass  
    print(i)
```
