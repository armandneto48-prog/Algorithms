```python
# ex4: 

# Versão 1:  (IA):

def quantos(V, x):
    me = 0  # Contador para menores
    ma = 0  # Contador para maiores
    
    for elemento in V:
        if elemento < x:
            me += 1
        elif elemento > x:
            ma += 1
            
    return me, ma

# Teste do exemplo do enunciado
Me, Ma = quantos([5, 1, 2, 3, 4], 3)
print(f"Me = {Me}, Ma = {Ma}")  # Saída: Me = 2, Ma = 2

# Versão 2:  (IA):

def quantos_pythonico(V, x):
    # Conta quantos elementos são menores que x
    me = sum(1 for elemento in V if elemento < x)
    
    # Conta quantos elementos são maiores que x
    ma = sum(1 for elemento in V if elemento > x)
    
    return me, ma

# Teste do exemplo do enunciado
Me, Ma = quantos_pythonico([5, 1, 2, 3, 4], 3)
print(f"Me = {Me}, Ma = {Ma}")  # Saída: Me = 2, Ma = 2
```

```python
# ex5:

# Eficiência Máxima:

def elementos_distintos_eficiente(V):
    # Um set elimina duplicados automaticamente, mas NÃO garante a ordem original
    resultado_set = set(V)
    return list(resultado_set)

# Exemplo de teste:
array_original = [4, 5, 1, 4, 2, 5, 3]
print(elementos_distintos_eficiente(array_original))
# Saída possível: [1, 2, 3, 4, 5] (a ordem pode mudar)

# Preservando a ordem original:

def elementos_distintos_ordenado(V):
    vistos = set()       # Guarda o que já encontrámos (rápido para verificar)
    resultado = []       # Guarda os elementos na ordem correta
    
    for elemento in V:
        if elemento not in vistos:
            vistos.add(elemento)
            resultado.append(elemento)
            
    return resultado

# Exemplo de teste:
array_original = [4, 5, 1, 4, 2, 5, 3]
print(elementos_distintos_ordenado(array_original))
# Saída garantida: [4, 5, 1, 2, 3] (ordem original preservada!)
```

```python
# ex6:

# Abordagem Iterativa:

def kaprekar_iterativo(num):
    # Garante que começamos com uma string de 4 algarismos (ex: '0378')
    num_str = str(num).zfill(4)
    
    iteracao = 0
    print(f"Número inicial: {num_str}")
    
    # O loop para assim que atingimos o ponto fixo 6174
    while num_str != "6174":
        iteracao += 1
        
        # 1. Ordena os algarismos para criar o maior e o menor número
        algarismos = list(num_str)
        maior = int("".join(sorted(algarismos, reverse=True)))
        menor = int("".join(sorted(algarismos)))
        
        # 2. Calcula a diferença
        resultado = maior - menor
        print(f"Iteração {iteracao}: {maior} - {menor:04d} = {resultado}")
        
        # 3. Atualiza a string para a próxima ronda, garantindo os 4 dígitos
        num_str = str(resultado).zfill(4)
        
    return iteracao

# Teste com o exemplo do enunciado:
print("--- TESTE ITERATIVO ---")
total_passos = kaprekar_iterativo(4321)
print(f"Resolvido em {total_passos} passos!\n")

# Abordagem Recursiva:

def kaprekar_recursivo(num, passo=0):
    # Transforma em string com 4 dígitos
    num_str = str(num).zfill(4)
    
    # Caso Base: Se chegou à constante, para a recursão e devolve os passos
    if num_str == "6174":
        return passo
        
    # Processamento dos algarismos
    algarismos = list(num_str)
    maior = int("".join(sorted(algarismos, reverse=True)))
    menor = int("".join(sorted(algarismos)))
    resultado = maior - menor
    
    print(f"Passo {passo + 1}: {maior} - {menor:04d} = {resultado}")
    
    # Chamada Recursiva: Passa o resultado e incrementa o contador de passos
    return kaprekar_recursivo(resultado, passo + 1)

# Teste com o exemplo do enunciado:
print("--- TESTE RECURSIVO ---")
total_passos_rec = kaprekar_recursivo(4321)
print(f"Resolvido em {total_passos_rec} passos!")
```
