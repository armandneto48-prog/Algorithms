```python
import numpy as np

# Configura o gerador moderno do NumPy (boas práticas)
rng = np.random.default_rng()

# Parâmetros da Beta (Alpha e Beta)
alpha, beta = 2.0, 5.0
tamanho_array = 10

# 1. Gera amostras contínuas entre 0 e 1
amostras_continuas = rng.beta(alpha, beta, size=tamanho_array)

# 2. Escala para o intervalo desejado (ex: inteiros de 0 a 100)
valores_maximos = 100
array_inteiros = (amostras_continuas * valores_maximos).astype(int)

print(array_inteiros)
# Resultado: Um array de inteiros concentrado mais perto de 0 (visto que alpha < beta)
```
