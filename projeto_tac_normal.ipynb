```python
# Pelos PowerPoints do Professor:

#Insertion Sort:

def insertionSort (V):
    N = len(V)
    for j in range(1,N):
        key = V[j]
        i = j - 1
    while (i >= 0 and V[i] > key):
        V[i+1] = V[i]
        i -= 1
        V[i+1] = key
```

```python
# BubbleSort (pelo código do professor):

def bubbleSort (V):
    j=0
    ok=False
    N = len(V)
    while (not ok):
        ok = True
        for i in range(N-1,j,-1):
            if (V[i-1] > V[i]):
                #swap
                V[i-1], V[i] = V[i], V[i-1] 
                ok = False
        j += 1
```

```python
# ShellSort (pelo código do professor):

def shellSort (V, gaps):
    N = len(V)
    for h in gaps:
        for i in range(h, N):
            key = V[i]
            j = i
            while( j >= h and V[j - h] > key ):
                V[j] = V[j - h]
                j -= h
            V[j] = key
```

```python
# 2º QuickSort Standard de Lomuto (versão mais detalhada):

def partition (V, i, f):
pivot = V[i] # pivot no leftmost
a = i # idx do maior valor
# atravessar array e comparar com pivot
for b in range(i+1, f+1):
if V[b] <= pivot:
# se encontrar valor < pivot trocar com
# valor referido por a
a = a + 1
V[a], V[b] = V[b], V[a]
# trocar pivot com valor referido por a
(V[a], V[i]) = (V[i], V[a])
return a # retorna pos do pivot
```

```python
# 1º QuickSort Standard de Lomuto (versão simples):
def quickSort(V, i, f):
    if(i < f):
        pivot = partition(V, i, f)
        quickSort(V, i, pivot - 1)
        quickSort(V, pivot + 1, f)
```

```python
# 2º QuickSort Standard de Lomuto (versão da partição):
def partition(V, i, f):
    pivot = V[i]  # pivot no leftmost
    a = i         # índice que rastreia os elementos menores
    
    # atravessar array e comparar com pivot
    for b in range(i + 1, f + 1):
        if V[b] <= pivot:
            # se encontrar valor <= pivot, incrementa 'a' e troca
            a = a + 1
            V[a], V[b] = V[b], V[a]
            
    # colocar o pivot na sua posição final correta (índice a)
    V[a], V[i] = V[i], V[a]
    return a # retorna a posição final do pivot
```

```python
# 1º QuickSort Standard de Hoare (versão simples):
def quickSort(V, i, f):
    if(i < f):
        pivot = partition2(V, i, f)
        quickSort(V, i, pivot - 1)
        quickSort(V, pivot + 1, f)
```

```python
# 2º QuickSort Standard de Hoare (versão da partição):
def partition2(V, i, f):
    p = V[i] # pivot no leftmost
    a = i
    b = f + 1
    
    while(True):
        # Avança o ponteiro da esquerda enquanto o valor for menor que o pivô
        while(True):
            a += 1
            if(a > f or V[a] >= p): 
                break 
        
        # Recua o ponteiro da direita enquanto o valor for maior que o pivô
        while(True):
            b -= 1
            if (b < i or V[b] <= p): 
                break 
        
        # Se os ponteiros se cruzarem, a partição terminou
        if(a >= b): 
            break 
            
        # Troca os elementos que estão nos lados errados
        V[a], V[b] = V[b], V[a]
        
    # Coloca o pivô na posição de separação (índice b)
    V[i], V[b] = V[b], V[i]
    return b # retorna a posição do pivot
```

```python
# QuickSort de 3 partições:

def partition3(A, i, f): 
    
    menors = i    
    x = i        
    maiors = f    
    pivot = A[i]  

    while (x <= maiors):
        if (A[x] < pivot):
            
            A[menors], A[x] = A[x], A[menors]
            menors += 1
            x += 1
        elif (A[x] > pivot):
           
            A[x], A[maiors] = A[maiors], A[x]
            maiors -= 1
        else:
            
            x += 1
            
    return menors, maiors 
```

```python
# CountingSort (pelo código do professor):


def countingSort(A, N, k):   # assume-se que A[1..N]
    # faz array B[]
    B = [0]
    for i in range(1,N+1):
        B += [0]
    C = [0]
    for i in range(1,k+1):
        C += [0]

    for i in range(1,N+1):
        C[A[i]] += 1
    for i in range(1,k+1):  # acumulado!
        C[i] += C[i-1]
    for j in range(N,0,-1):
        B[C[A[j]]] = A[j]
        C[A[j]] -= 1

    return B
```
