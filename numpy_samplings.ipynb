```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate data
data = np.random.normal(loc=0, scale=1, size=1000)

# Plot histogram
plt.hist(data, bins=30, edgecolor='black', density=True)

# Plot theoretical PDF
x = np.linspace(-4, 4, 200)
pdf = norm.pdf(x, loc=0, scale=1)
plt.plot(x, pdf, label="Theoretical PDF")

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Examples:

# Generate a uniform random sample from np.arange(5) of size 3:

rng = np.random.default_rng()
rng.choice(5, 3)

#This is equivalent to rng.integers(0,5,3)

# Generate a non-uniform random sample from np.arange(5) of size 3:

rng.choice(5, 3, p=[0.1, 0, 0.3, 0.6, 0])

# Generate a uniform random sample from np.arange(5) of size 3 without replacement:

rng.choice(5, 3, replace=False)

#This is equivalent to rng.permutation(np.arange(5))[:3]

# Generate a uniform random sample from a 2-D array along the first axis (the default), without replacement:

rng.choice([[0, 1, 2], [3, 4, 5], [6, 7, 8]], 2, replace=False)

# Generate a non-uniform random sample from np.arange(5) of size 3 without replacement:

rng.choice(5, 3, replace=False, p=[0.1, 0, 0.3, 0.6, 0])

# Generate a non-uniform random sample from np.arange(5) of size 3 without replacement:

rng.choice(5, 3, replace=False, p=[0.1, 0, 0.3, 0.6, 0])

# Generate a non-uniform random sample from np.arange(5) of size 3 without replacement:

rng.choice(5, 3, replace=False, p=[0.1, 0, 0.3, 0.6, 0])
```

```python
import numpy as np

rng = np.random.default_rng()

print("=== INÍCIO DA PARAMETRIZAÇÃO: DISTRIBUIÇÃO BETA - COM REPOSIÇÃO ===")

# Parâmetros 'a' e 'b' (alfa e beta). Devem ser maiores que 0.
populacao_beta = rng.beta(a=2, b=5, size=10000)

tamanhos = np.linspace(30, 10000, num=15, dtype=int)

for N in tamanhos:
    # 1. Fazemos a amostragem com reposição
    amostra_continua = rng.choice(populacao_beta, size=N, replace=True)
    
    # 2. DISCRETIZAÇÃO: Multiplicamos por 100 e arredondamos para inteiros
    amostra_experimental = np.round(amostra_continua * 100).astype(int)
    
    print(f"\n[Beta] Amostra para N = {N}:")
    print(amostra_experimental)
    print(f"-> Tamanho real do array: {len(amostra_experimental)}")
```

```python
import numpy as np

rng = np.random.default_rng()

print("\n=== INÍCIO DA PARAMETRIZAÇÃO: DISTRIBUIÇÃO POISSON - COM REPOSIÇÃO ===")

# Parâmetro 'lam' (lambda = média de eventos, ex: 4)
populacao_poisson = rng.poisson(lam=4, size=10000)

tamanhos = np.linspace(30, 10000, num=15, dtype=int)

for N in tamanhos:
    # Como a população já é de inteiros, basta fazer o choice direto
    amostra_experimental = rng.choice(populacao_poisson, size=N, replace=True)
    
    print(f"\n[Poisson] Amostra para N = {N}:")
    print(amostra_experimental)
    print(f"-> Tamanho real do array: {len(amostra_experimental)}")
```

```python
import numpy as np

rng = np.random.default_rng()

rng.standard_normal()

s = rng.standard_normal(10)

print(s)

# Até aqui: geração de dados aleatórios mediante uma distribuição Normal.

# A seguir, tentaremos recorrer a um processo de discretização dos dados gerados:

# Segundo passo: Multiplicação por um fator de escala (broadcasting):

ten = 10 * s

print(ten)

# Terceiro passo: Rounding:

r = round(56294995.3421315, 3)

print("Primeiro arredondamento: " + str(r))

print("Próximos arredondamentos: ")

np.round(16.055, 2), round(16.055, 2)  # equals 16.0549999999999997
```
