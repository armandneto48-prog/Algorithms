```python
# Example of time library in BubbleSort array:

import time
import numpy as np

# 1. Definição do Algoritmo Bubble Sort
def bubbleSort(arr):
    n = len(arr)
    # Percorre todos os elementos do array
    for i in range(n):
        # Os últimos i elementos já estão ordenados, não precisamos de os rever
        for j in range(0, n - i - 1):
            # Troca se o elemento encontrado for maior que o próximo elemento
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

# 2. Gerar um dataset de teste desordenado (usando NumPy)
# Vamos criar um array de 1000 elementos com uma distribuição uniforme
array_original = np.random.randint(0, 5000, size=1000)

# 3. Preparação para o Benchmark
# Criamos uma cópia profunda para não estragar o array original
array_para_ordenar = array_original.copy()

# --- INÍCIO DA MEDIÇÃO ---
inicio = time.perf_counter()

# O algoritmo de ordenação executa aqui:
bubbleSort(array_para_ordenar)

fim = time.perf_counter()
# --- FIM DA MEDIÇÃO ---

tempo_decorrido = fim - inicio

# 4. Demonstração de resultados
print(f"Tamanho do array: {len(array_original)} elementos.")
print(f"O Bubble Sort demorou {tempo_decorrido:.6f} segundos a ordenar.")
```
