```python
# ShellSort (pelo código do professor):

def shellSort (V, gaps):
    N = len(V)
    for h in gaps:
        for i in range(h, N):
            key = V[i]
            j = i
            while( j >= h and V[j - h] > key ):
                V[j] = V[j - h]
                j -= h
            V[j] = key
```

```python
import numpy as np
import time

# --- 1. ALGORITMO DO PROFESSOR (ShellSort) ---
def shellSort(V, gaps):
    N = len(V)
    for h in gaps:
        for i in range(h, N):
            key = V[i]
            j = i
            while j >= h and V[j - h] > key:
                V[j] = V[j - h]
                j -= h
            V[j] = key

# --- 2. CONFIGURAÇÃO DO BENCHMARKING ---
rng = np.random.default_rng()

print("=== INÍCIO DO BENCHMARKING: NORMAL + SHELL SORT ===")

populacao_normal = rng.normal(loc=10, scale=2, size=10000)
tamanhos = np.linspace(30, 10000, num=15, dtype=int)

# --- 3. LOOP DE EXECUÇÃO E MEDIÇÃO ---
for N in tamanhos:
    print(f"\nA processar N = {N} com Shell Sort...")
    
    # 1. Geramos a sequência de Gaps padrão para este tamanho N específico
    gaps_sequencia = []
    h = N // 2
    while h > 0:
        gaps_sequencia.append(h)
        h //= 2
        
    # 2. Geramos a amostra experimental com reposição
    amostra_experimental = rng.choice(populacao_normal, size=N, replace=True)
    
    # CRONOMETRAGEM
    tempo_inicial = time.perf_counter()
    
    # Execução do algoritmo passando a amostra e a lista de gaps gerada
    shellSort(amostra_experimental, gaps_sequencia)
    
    tempo_final = time.perf_counter()
    tempo_execucao = tempo_final - tempo_inicial
    
    print(f"-> [Sucesso] Array de tamanho {N} ordenado pelo Shell Sort.")
    print(f"-> Tempo de execução: {tempo_execucao:.6f} segundos")
```

```python
import numpy as np
import time
import sys

# Garante que o Python aguenta a profundidade da recursão
sys.setrecursionlimit(20000)

# --- 1. ALGORITMO (QuickSort de Hoare + Partição2) ---
def partition2(V, i, f):
    pivot = V[i]  # Escolha clássica do pivot mais à esquerda
    esquerda = i - 1
    direita = f + 1
    
    while True:
        # Avança o ponteiro da esquerda enquanto os valores forem menores
        esquerda += 1
        while V[esquerda] < pivot:
            esquerda += 1
            
        # Recua o ponteiro da direita enquanto os valores forem maiores
        direita -= 1
        while V[direita] > pivot:
            direita -= 1
            
        # Se os ponteiros se cruzarem, a partição terminou
        if esquerda >= direita:
            return direita
            
        # Caso contrário, troca os elementos desalinhados
        V[esquerda], V[direita] = V[direita], V[esquerda]

def quickSortHoare(V, i, f):
    if i < f:
        # Nota: Na partição de Hoare clássica, o pivot divide o array em [i..pivot] e [pivot+1..f]
        pivot = partition2(V, i, f)
        quickSortHoare(V, i, pivot)
        quickSortHoare(V, pivot + 1, f)

# --- 2. CONFIGURAÇÃO DO BENCHMARKING ---
rng = np.random.default_rng()

print("=== INÍCIO DO BENCHMARKING: NORMAL + QUICKSORT (HOARE) ===")

populacao_normal = rng.normal(loc=10, scale=2, size=10000)
tamanhos = np.linspace(30, 10000, num=15, dtype=int)

# --- 3. LOOP DE EXECUÇÃO E MEDIÇÃO ---
for N in tamanhos:
    print(f"\nA processar N = {N} com QuickSort Hoare...")
    
    # Amostragem experimental com reposição
    amostra_experimental = rng.choice(populacao_normal, size=N, replace=True)
    
    # CRONOMETRAGEM
    tempo_inicial = time.perf_counter()
    
    # Execução passando os índices de início e fim
    quickSortHoare(amostra_experimental, 0, N - 1)
    
    tempo_final = time.perf_counter()
    tempo_execucao = tempo_final - tempo_inicial
    
    print(f"-> [Sucesso] Array de tamanho {N} ordenado por Hoare.")
    print(f"-> Tempo de execução: {tempo_execucao:.6f} segundos")
```

```python
# 2º QuickSort Standard de Hoare (versão da partição):
def partition2(V, i, f):
    p = V[i] # pivot no leftmost
    a = i
    b = f + 1
    
    while(True):
        # Avança o ponteiro da esquerda enquanto o valor for menor que o pivô
        while(True):
            a += 1
            if(a > f or V[a] >= p): 
                break 
        
        # Recua o ponteiro da direita enquanto o valor for maior que o pivô
        while(True):
            b -= 1
            if (b < i or V[b] <= p): 
                break 
        
        # Se os ponteiros se cruzarem, a partição terminou
        if(a >= b): 
            break 
            
        # Troca os elementos que estão nos lados errados
        V[a], V[b] = V[b], V[a]
        
    # Coloca o pivô na posição de separação (índice b)
    V[i], V[b] = V[b], V[i]
    return b # retorna a posição do pivot
```

```python
import numpy as np
import time
import sys

# Distribuição Discreta Sem Reposição:

# Ajuste de segurança para a pilha recursiva dos QuickSorts
sys.setrecursionlimit(30000)

# =====================================================================
# 1. DEFINIÇÃO DOS ALGORITMOS DE ACORDO COM AS REGRAS DE SALA DE AULA
# =====================================================================

def insertionSort(V):
    N = len(V)
    for j in range(1, N):
        key = V[j]
        i = j - 1
        while i >= 0 and V[i] > key:
            V[i + 1] = V[i]
            i -= 1
        V[i + 1] = key

def bubbleSort(V):
    j = 0
    ok = False
    N = len(V)
    while not ok:
        ok = True
        for i in range(N - 1, j, -1):
            if V[i - 1] > V[i]:
                V[i - 1], V[i] = V[i], V[i - 1]
                ok = False
        j += 1

def shellSort(V, gaps):
    N = len(V)
    for h in gaps:
        for i in range(h, N):
            key = V[i]
            j = i
            while j >= h and V[j - h] > key:
                V[j] = V[j - h]
                j -= h
            V[j] = key

# --- VARIANTES DO QUICKSORT LOMUTO ---
def partitionLomuto(V, i, f):
    pivot = V[i]
    a = i
    for b in range(i + 1, f + 1):
        if V[b] <= pivot:
            a = a + 1
            V[a], V[b] = V[b], V[a]
    V[a], V[i] = V[i], V[a]
    return a

def quickSortLomuto(V, i, f):
    if i < f:
        pivot = partitionLomuto(V, i, f)
        quickSortLomuto(V, i, pivot - 1)
        quickSortLomuto(V, pivot + 1, f)

# --- VARIANTES DO QUICKSORT HOARE ---
def partitionHoare(V, i, f):
    p = V[i]
    a = i - 1    
    b = f + 1    
    while True:
        while True:
            a += 1
            if V[a] >= p: 
                break 
        while True:
            b -= 1
            if V[b] <= p: 
                break 
        if a >= b: 
            return b
        V[a], V[b] = V[b], V[a]

def quickSortHoare(V, i, f):
    if i < f:
        pivot = partitionHoare(V, i, f)
        if pivot == f:
            quickSortHoare(V, i, pivot - 1)
        else:
            quickSortHoare(V, i, pivot)
            quickSortHoare(V, pivot + 1, f)

# --- QUICKSORT 3 PARTIÇÕES ---
def partition3(A, i, f): 
    menors = i    
    x = i        
    maiors = f    
    pivot = A[i]  
    while x <= maiors:
        if A[x] < pivot:
            A[menors], A[x] = A[x], A[menors]
            menors += 1
            x += 1
        elif A[x] > pivot:
            A[x], A[maiors] = A[maiors], A[x]
            maiors -= 1
        else:
            x += 1
    return menors, maiors

def quickSort3Particoes(A, i, f):
    if i < f:
        esq, dir_idx = partition3(A, i, f)
        quickSort3Particoes(A, i, esq - 1)
        quickSort3Particoes(A, dir_idx + 1, f)

# --- COUNTING SORT ---
def countingSort(A, N, k):
    B = [0]
    for i in range(1, N + 1):
        B += [0]
    C = [0]
    for i in range(1, k + 1):
        C += [0]

    for i in range(1, N + 1):
        C[A[i]] += 1
    for i in range(1, k + 1):
        C[i] += C[i - 1]
    for j in range(N, 0, -1):
        B[C[A[j]]] = A[j]
        C[A[j]] -= 1
    return B

# =====================================================================
# 2. CONFIGURAÇÃO DA AMOSTRAGEM (DISTRIBUIÇÃO UNIFORME DISCRETA)
# =====================================================================
rng = np.random.default_rng()

print("=== INÍCIO DA PARAMETRIZAÇÃO: DISTRIBUIÇÃO UNIFORME DISCRETA - SEM REPOSIÇÃO ===")
# Intervalo de 1 a 500
populacao_uniforme = rng.integers(low=1, high=501, size=10000)
tamanhos = np.linspace(30, 10000, num=15, dtype=int)

# =====================================================================
# 3. EXECUÇÃO DO BENCHMARK EM LOOP
# =====================================================================
for N in tamanhos:
    print(f"\n==================================================")
    print(f"PROCESSANDO TAMANHO DE AMOSTRA N = {N}")
    print(f"==================================================")
    
    # Amostragem com reposição direta
    amostra_base = rng.choice(populacao_uniforme, size=N, replace=False)
    
    # --- TESTE 1: Insertion Sort ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    insertionSort(v_test)
    print(f"-> Insertion Sort:          {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 2: Bubble Sort ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    bubbleSort(v_test)
    print(f"-> Bubble Sort:             {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 3: Shell Sort ---
    v_test = amostra_base.copy()
    gaps_seq = []
    h = N // 2
    while h > 0:
        gaps_seq.append(h)
        h //= 2
    t_ini = time.perf_counter()
    shellSort(v_test, gaps_seq)
    print(f"-> Shell Sort:              {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 4: QuickSort Lomuto Completo ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    quickSortLomuto(v_test, 0, N - 1)
    print(f"-> QS Lomuto Completo:      {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 5: Partição Lomuto Isolada ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    partitionLomuto(v_test, 0, N - 1)
    print(f"-> Partição Lomuto Única:   {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 6: QuickSort Hoare Completo ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    quickSortHoare(v_test, 0, N - 1)
    print(f"-> QS Hoare Completo:       {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 7: Partição Hoare Isolada ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    partitionHoare(v_test, 0, N - 1)
    print(f"-> Partição Hoare Única:    {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 8: QuickSort 3 Partições ---
    v_test = amostra_base.copy()
    t_ini = time.perf_counter()
    quickSort3Particoes(v_test, 0, N - 1)
    print(f"-> QS 3 Partições:          {time.perf_counter() - t_ini:.6f} seg")
    
    # --- TESTE 9: Counting Sort ---
    k_max = int(np.max(amostra_base))
    v_offset = [0] + list(amostra_base)
    t_ini = time.perf_counter()
    _ = countingSort(v_offset, N, k_max)
    print(f"-> Counting Sort:           {time.perf_counter() - t_ini:.6f} seg")
```
