```python
# Upper Bound Limit Through Theorem of Lamé

def mdc (ma, me):
    while(me != 0):
        t = me
        me = ma % me
        ma = t
    return ma

a = mdc(40,5)

print(a)
```

```python
def mdc(a, b):
    while b:
        a, b = b, a % b
    return a

# Teste
print(mdc(48, 18)) # Saída: 6
```

```python
def procura(x, v):
    i = 0
    N = len(V)
    while (i < N and v[i] != x):
        i += 1
    if (i == N):
        return - 1
    else:
        return i
    
# Teste
V = [1, 2, 3, 4, 5]
print(procura(3, V)) # Saída: 2
```

```python
# Representação do Grafo: Tesoura -> Papel -> Pedra -> Tesoura
# Cada chave é um "Nó" (Node) e o valor é a lista de "Arestas" (Edges) que saem dele.

grafo_jogo = {
    "Tesoura": ["Papel"],
    "Papel": ["Pedra"],
    "Pedra": ["Tesoura"]
}

def verificar_vitoria(jogador1, jogador2):
    # Verifica se existe uma aresta direcionada de jogador1 para jogador2
    if jogador2 in grafo_jogo.get(jogador1, []):
        return f"{jogador1} ganha de {jogador2}!"
    elif jogador1 in grafo_jogo.get(jogador2, []):
        return f"{jogador2} ganha de {jogador1}!"
    else:
        return "Empate ou jogada inválida."

# Teste do Ciclo
print("--- Simulação de Grafo Direcionado ---")
print(verificar_vitoria("Tesoura", "Papel")) # Caminho: Tesoura -> Papel
print(verificar_vitoria("Papel", "Pedra"))   # Caminho: Papel -> Pedra
print(verificar_vitoria("Pedra", "Tesoura")) # Caminho: Pedra -> Tesoura (Fechando o Ciclo)
```
